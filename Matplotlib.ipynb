3D Surface_Plot

```python
x = np.linspace(-10,10,20)
y = np.linspace(-10,10,20)
X,Y = np.meshgrid(x,y)
z = X**2 + Y**2
fig = plt.figure(figsize=(8,6),dpi=100)
ax = plt.subplot(projection ="3d")
p = ax.plot_surface(X,Y,z,cmap = "twilight") #cmap is for gradient of color
fig.colorbar(p)
```

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
X,Y = np.meshgrid(x,y)
z = np.cos(Y) + np.sin(Y)
fig = plt.figure(figsize=(8,6),dpi=100)
ax = plt.subplot(projection ="3d")
p = ax.plot_surface(X,Y,z,cmap = 'viridis') #cmap is for gradient of color
fig.colorbar(p)
plt.show()
```

```python
x = np.linspace(-10,10,20)
y = np.linspace(-10,10,20)
X,Y = np.meshgrid(x,y)
z = X**2 + Y**2
fig = plt.figure(figsize=(8,6),dpi=100)
ax = plt.subplot()
p = ax.contour(X,Y,z,cmap = "twilight") #cmap is for gradient of color
fig.colorbar(p)
```

```python
x = np.linspace(-10,10,20)
y = np.linspace(-10,10,20)
X,Y = np.meshgrid(x,y)
z = X**2 + Y**2
fig = plt.figure(figsize=(8,6),dpi=100)
ax = plt.subplot()
p = ax.contourf(X,Y,z,cmap = "twilight") #cmap is for gradient of color
fig.colorbar(p)
```

```python
fig = plt.figure()
ax1 = fig.add_subplot(2, 1, 1, projection="3d")
ax2 = fig.add_subplot(2, 1, 2)

ax1.plot_surface(X, Y, z)
ax2.contourf(X, Y, z)

plt.show()
```
