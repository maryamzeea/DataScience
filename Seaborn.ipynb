```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = sns.load_dataset("titanic")
sns.set_style("whitegrid")
```

```python
titanic
```

```python
sns.jointplot(data=titanic, x="fare", y="age")
```

```python
sns.countplot(data=titanic,x="fare", hue="survived")
```

```python
sns.boxplot(data=titanic, x="class", y="age",hue="survived")
```

```python
sns.stripplot(data = titanic, x="class", y="age")
```

```python
sns.countplot(data=titanic, x="sex", hue="survived")
```

```python
age = titanic.pivot_table(
    index="age",
    columns="class",
    values="survived",
    aggfunc="mean"
)
sns.heatmap(age)
```

```python
g = sns.PairGrid(titanic)
g.map(sns.scatterplot)
```

```python
sns.displot(data=titanic,x='age',col = "sex",kind='hist',hue="survived")
```

```python
sns.heatmap(titanic.corr(numeric_only=True),cmap='rainbow')
```
